```python

import openpyxl
import pandas as pd

# =========================
# 1. LOAD FILE
# =========================
file_path = "downloads/RGB-T-Verified.xlsx"
df = pd.read_excel(file_path, engine="openpyxl")

print("Fichier chargé ✅")
print("Total articles :", len(df))

# =========================
# 2. CHECK COLUMN
# =========================
score_col = "Final Score"

if score_col not in df.columns:
    raise ValueError(f"La colonne '{score_col}' n'existe pas dans le fichier.")

# convertir en numérique au cas où
df[score_col] = pd.to_numeric(df[score_col], errors="coerce")

# =========================
# 3. NEW DECISION BASED ONLY ON FINAL SCORE
# =========================
df["Decision_by_FinalScore"] = df[score_col].apply(
    lambda x: "Include" if pd.notna(x) and x >= 4 else "Exclude"
)

# =========================
# 4. COUNT RESULTS
# =========================
include_count = (df["Decision_by_FinalScore"] == "Include").sum()
exclude_count = (df["Decision_by_FinalScore"] == "Exclude").sum()

print("\nRésultats basés sur Final Score :")
print("Include :", include_count)
print("Exclude :", exclude_count)



# =========================
# 5. SAVE ONLY INCLUDED ARTICLES
# =========================
articles_retenus = df[df["Decision_by_FinalScore"] == "Include"]

output_file_includes = "downloads/articles-retenus.xlsx"
articles_retenus.to_excel(output_file_includes, index=False)


print("Nombre d'articles retenus :", len(articles_retenus))
```
