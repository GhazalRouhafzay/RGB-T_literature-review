```python
import pandas as pd
import re
from collections import Counter

# Charger fichier
df = pd.read_excel("downloads/articles-retenus-final.xlsx")

# =========================
# TABLEAU 1 : Techniques
# =========================

all_techniques = []

for tech in df["Techniques Used"].dropna():
    parts = re.split(r",|;", str(tech))
    parts = [p.strip().lower() for p in parts if p.strip() != ""]
    all_techniques.extend(parts)

tech_count = Counter(all_techniques)

tech_df = pd.DataFrame(tech_count.items(), columns=["Technique", "Count"])
tech_df = tech_df.sort_values(by="Count", ascending=False)

print("\n=== Techniques ===")
print(tech_df)

# =========================
# TABLEAU 2 : Learning Type
# =========================

all_learning = []

for lt in df["Learning Type"].dropna():
    parts = re.split(r",|;", str(lt))
    parts = [p.strip().lower() for p in parts if p.strip() != ""]
    all_learning.extend(parts)

learning_count = Counter(all_learning)

learning_df = pd.DataFrame(learning_count.items(), columns=["Learning Type", "Count"])
learning_df["Percentage"] = (learning_df["Count"] / len(df)) * 100
learning_df = learning_df.sort_values(by="Percentage", ascending=False)

print("\n=== Learning Types ===")
print(learning_df)
```
